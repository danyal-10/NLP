# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

## Task 2 — Encoder-only Model

Use the sentiment classifier on these three sentences:

```text
The movie was excellent.
The lecture was difficult to understand.
I enjoyed learning about transformers.
```

Print the predicted label and score for each sentence.

````python
from transformers import pipeline

classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

sentences = [
    "The movie was excellent.",
    "The lecture was difficult to understand.",
    "I enjoyed learning about transformers."
]

results = classifier(sentences)

for sentence, result in zip(sentences, results):
    print(f"Sentence : {sentence}")
    print(f"Label    : {result['label']}")
    print(f"Score    : {result['score']:.4f}")
    print("-" * 45)
````

**Explanation:**
- *"The movie was excellent."* → **POSITIVE**, because the word *excellent* clearly expresses a positive opinion.
- *"The lecture was difficult to understand."* → **NEGATIVE**, because *difficult to understand* describes a bad experience.
- *"I enjoyed learning about transformers."* → **POSITIVE**, because *enjoyed* expresses a positive feeling.
- The **score** is the model's confidence (probability) in the predicted label; a score close to 1 means the model is very sure.

## Task 3 — Encoder–Decoder Model

Use the FLAN-T5 model to perform the following:

- Summarize a short paragraph.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

paragraph = (
    "Natural language processing is a field of artificial intelligence that helps computers "
    "understand, interpret and generate human language. It is used in many applications such as "
    "machine translation, chatbots, sentiment analysis and search engines. In recent years, "
    "transformer models like BERT and GPT have greatly improved the performance of NLP systems "
    "because they can learn the meaning of words from their context in very large amounts of text."
)

prompt = "Summarize: " + paragraph

# Convert text into tokens
inputs = tokenizer(prompt, return_tensors="pt")

# Generate the summary
output = model.generate(**inputs, max_new_tokens=60, num_beams=4)

# Convert tokens back to text
summary = tokenizer.decode(output[0], skip_special_tokens=True)

print("Original paragraph:\n", paragraph)
print("\nOriginal length:", len(paragraph.split()), "words")
print("\nSummary:\n", summary)
print("\nSummary length:", len(summary.split()), "words")
````

**Explanation:**
- FLAN-T5 is an **encoder–decoder** model: the **encoder** reads the full paragraph and the **decoder** writes a new, shorter text.
- FLAN-T5 was trained on many instructions, so we only need to add the instruction *"Summarize:"* at the start of the input to tell it which task to perform.
- `num_beams=4` (beam search) makes the model compare several possible outputs and keep the best one, and `max_new_tokens=60` limits the length of the summary.
- FLAN-T5-*small* is a very small model, so its summary is short and simple; larger versions (flan-t5-base / large) produce better summaries.


## Task 5 — Short Reflection

Answer in 3–5 sentences:

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


### ✅ Answer — Task 5

A decoder-only model is trained to **predict the next token** based on all the previous tokens, which is exactly what a chatbot does when it writes a reply word by word. In a conversation, the whole chat history (the user's messages and the model's previous answers) is simply treated as one long text, and the model continues it by generating the next answer. Because it uses **causal (masked) self-attention**, each new token can look at everything that came before it, so the model keeps the context of the conversation. Decoder-only models are also easy to train on huge amounts of plain text from the internet and scale very well, and after fine-tuning on instructions and human feedback they learn to follow questions and answer them in a helpful way. That is why models like ChatGPT are built on the decoder-only GPT architecture.

### Question

Why do you sometimes get different outputs even when the prompt is the same?

**Hint:** Sampling introduces randomness when choosing the next token.


````python
from transformers import pipeline, set_seed

generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

# With a fixed seed: the result is the same every time this cell is run
set_seed(42)
output = generator(prompt, max_new_tokens=30, do_sample=True, top_k=50, temperature=0.9)
print("With set_seed(42):\n", output[0]["generated_text"])
````

````python
# Without set_seed: run the same prompt 3 times
for i in range(3):
    output = generator(prompt, max_new_tokens=30, do_sample=True, top_k=50, temperature=0.9)
    print(f"Run {i + 1}:", output[0]["generated_text"])
    print("-" * 60)
````

### ✅ Answer — Optional Challenge

**Why do we get different outputs for the same prompt?**

With `do_sample=True` the model does **not** always pick the most likely next token. Instead, it computes a probability for every possible next token and then **randomly samples** one of them according to these probabilities (here only from the 50 most likely tokens, `top_k=50`). Because a random choice is made at every step, the first different token changes everything that comes after it, so each run produces a different text.

`set_seed(42)` fixes the random number generator, so the "random" choices are the same every time and the output is repeatable. With `do_sample=False` (greedy decoding, used in Part A) the model always picks the most probable token, so the output is always the same even without a seed.

The `temperature` controls how random the sampling is: a low temperature makes the output more predictable, and a high temperature makes it more creative and varied.

## What I Learned from This Lab

- **Large Language Models** are built on the **Transformer** architecture, and there are three main families: **encoder-only** (BERT), **encoder–decoder** (T5, BART) and **decoder-only** (GPT).
- **Choosing the right family depends on the task:** encoder-only models are best for understanding and classifying text (sentiment, NER), encoder–decoder models for turning one text into another (translation, summarization), and decoder-only models for generating text (chatbots, completion).
- **Hugging Face `transformers`:** the `pipeline()` function lets me use a pre-trained model in one line, and `AutoTokenizer` / `AutoModelForSeq2SeqLM` give more control: tokenize the text → `generate()` → decode the tokens back to text.
- **Pre-trained models can be used directly** without training them myself: I used DistilBERT for sentiment analysis and FLAN-T5 for summarization by just giving it an instruction like *"Summarize:"*.
- **The score** returned by a classifier is the model's confidence in its prediction.
- **How text is generated:** decoder-only models predict the next token one at a time. **Greedy decoding** always picks the most likely token, while **sampling** picks randomly, which is why the output changes every run unless I use `set_seed()`. Parameters like `temperature`, `top_k` and `num_beams` control how the text is generated.
- **Model size matters:** small models like DistilGPT-2 and FLAN-T5-small are fast, but their outputs are simpler and sometimes wrong (for example, the French translation in Part B was not correct), while bigger models give much better results.
